```python
def hourglass(n):
    for i in range(n,0,-1):
        k=2*n+3
        print(" "*(k-i),end="")
        print("*"*(2*i+1))
    for i in range(n+1):
        k=2*n+3
        print(" "*(k-i),end="")
        print("*"*(2*i+1))
    

hourglass(5)
```

```python
def semidia(n):
    for i in range(n):
        print("* "*i)
    for i in range(n,0,-1):
        print("* "*i)
    
semidia(5)
```

```python
def square(n):
    for i in range(n):
        if(i==0 or i==n-1):
            print("*"*n)
        else:
            # print("*",end="")
            # print(" "*(2*(n-2)+1),end="")
            # print("*")
            print("*"+" "*(n-2)+"*")

square(5)
```

```python
def triangle(n):
    print(" "*(n-1),"*")
    for i in range(1,n):
        print(f"{' '*(n-i)}*{' '*(2*i-1)}*")
    for i in range(n,0,-1):
        print(f"{' '*(n-i)}*{' '*(2*i-1)}*")
    print(" "*(n-1),"*")

triangle(5)
```

```python
def diamond(n):
    rows=2*n-1
    cols=2*n-1
    mid=n-1
    for i in range(rows):
        d=abs(i-mid)
        for j in range(cols):
            if(j==d or j==cols-1-d):
                print("*",end='')
            else:
                print(" ",end='')
        print()

diamond(5)
```

```python
def hollow_diamond(n):
    rows = 2 * n - 1
    cols = 2 * n - 1

    mid = n - 1

    for row in range(rows):
        # Distance from the middle row
        d = abs(mid - row)

        left = d
        right = cols - 1 - d

        for col in range(cols):
            if col == left or col == right:
                print("*", end="")
            else:
                print(" ", end="")
        print()

hollow_diamond(5)
```

```python
def floyd(n):
    for i in range(n):
        x=0
        for j in range(2*i+1):
            if(j<=i):
                x+=1
                print(x,end='')
            else:
                x-=1
                print(x,end='')
        print()

floyd(5)
```
