```python
def diamond(n):
    for i in range(n+1):
        k=n
        print(" "*(k-i),end="")
        print("* "*i)
    for i in range(n-1,0,-1):
        k=n
        print(" "*(k-i),end="")
        print("* "*i)

diamond(5)
```

```python
def hourglass(n):
    for i in range(n,0,-1):
        k=2*n+3
        print(" "*(k-i),end="")
        print("*"*(2*i+1))
    for i in range(n+1):
        k=2*n+3
        print(" "*(k-i),end="")
        print("*"*(2*i+1))
    

hourglass(5)
```

```python
def semidia(n):
    for i in range(n):
        print("* "*i)
    for i in range(n,0,-1):
        print("* "*i)
    
semidia(5)
```
