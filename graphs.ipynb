```python
n = 12

edges = [
    (0, 1), (0, 2), (0, 3),
    (1, 4), (1, 5),
    (2, 5), (2, 6),
    (3, 6), (3, 7),
    (4, 8), (5, 8), (5, 9),
    (6, 9), (6, 10),
    (7, 10), (7, 11),
    (8, 11), (9, 10)
]
graph=[[] for _ in range(n)]

for u,v in edges:
    graph[u].append(v)
    graph[v].append(u)

graph
```

### BFS Traversal

```python
from collections import deque

def bfs(graph,start):
    visited={start}
    q=deque([start])
    re=[]

    while q:
        node=q.popleft()
        re.append(node)
        
        for n in graph[node]:
            if n not in visited:
                q.append(n)
                visited.add(n)
    return re


print(bfs(graph,0))
```

```python
def dfs(graph,start):
    visited=set()
    result=[]

    def visit(node):
        result.append(node)
        visited.add(node)

        for n in graph[node]:
            if n not in visited:
                visit(n)

    visit(start)
    return result

print(dfs(graph,0))
```

```python
grid = [
    [1, 1, 0],
    [0, 1, 0],
    [1, 1, 1]
]

dir=[(0,-1),(0,1),(1,0),(-1,0)]

def bfs(r,c):
    rows=len(grid)
    cols=len(grid[0])
    q=deque([(r,c)])
    visited={(r,c)}
    result=[]

    while q:
        sr,sc=q.popleft()
        result.append((sr,sc))

        for dr,dc in dir:
            nr,nc=sr+dr,sc+dc

            if(
                0<=nr<rows and 0<=nc<cols
                and grid[nr][nc]==1 and (nr,nc) not in visited
            ):
                visited.add((nr,nc))
                q.append((nr,nc))
        
    return result


print(bfs(0,0))
```

```python
def dfs(r,c):
    res=[]
    visited=set()
    rows,cols=len(grid),len(grid[0])
    dir=[(0,-1),(0,1),(1,0),(-1,0)]

    def visit(r,c):
        visited.add((r,c))
        res.append((r,c))

        for  nr,nc in dir:
            dr,dc=r+nr,c+nc
            if (
                0<=dr<rows and 0<=dc<cols and grid[dr][dc]==1 and (dr,dc) not in visited
            ):
                visit(dr,dc)
    visit(r,c)
    return res

dfs(0,0)
```
